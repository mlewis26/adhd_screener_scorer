```python
people = [
    {"name": "Ana", "answers": [3, 2, 1, 4, 3, 2]},
    {"name": "Ben", "answers": [1, 1, 2, 2, 1, 0]},
    {"name": "Cy",  "answers": [4, 4, 3, 3, 4, 3]},
    {"name": "Dee", "answers": [2, 1, 2, 2, 3, 2]},]
cutoffs = [2, 2, 2, 3, 3, 3]

print(people[0]["answers"])
```

```python
count = 0
ana = (people[0]["answers"])
ben = (people[1]["answers"])
cy = (people[2]["answers"])
dee = (people[3]["answers"])
for i in range(6):
    if ana[i] >= cutoffs[i]:
        count = count + 1
print("Ana:", count)
if count >= 4:
    print("Positive screen")
else: 
    print("Negative screen")
count = 0
for i in range(6): 
    if ben[i] >= cutoffs[i]:
        count = count + 1
print("Ben:", count)
if count >= 4:
    print("Positive screen")
else:
    print("Negative screen")
count = 0 
for i in range(6): 
    if cy[i] >= cutoffs[i]:
        count = count + 1
print("Cy:", count)
if count >= 4:
    print("Positive screen")
else:
    print("Negative screen")
count = 0
for i in range(6): 
    if dee[i] >= cutoffs[i]:
        count = count + 1
print("Dee:", count)
if count >= 4:
    print("Positive screen")
else:
    print("Negative screen")
```

```python
for p in people:
    count = 0 
    answers = (p["answers"])
    for i in range(6):
        if answers[i] >= cutoffs[i]:
            count = count + 1
    print(p["name"], count)
    if count >= 4:
        print("Positive screen")
    else: 
        print("Negative screen")
```

```python
import csv
with open("asrs_data.csv") as f:
    reader = csv.DictReader(f)
    for row in reader:
        answers = [int(row["q1"]), int(row["q2"]), int(row["q3"]), int(row["q4"]), int(row["q5"]), int(row["q6"])]
        count = 0 
        for i in range(6):
            if answers[i] >= cutoffs[i]:
                count = count + 1
        print(row["name"], count)
        if count >= 4:
            print("Positive screen")
        else: 
            print("Negative screen")
```
