```python
people = [
    {"name": "Ana", "answers": [3, 2, 1, 4, 3, 2]},
    {"name": "Ben", "answers": [1, 1, 2, 2, 1, 0]},
    {"name": "Cy",  "answers": [4, 4, 3, 3, 4, 3]},
    {"name": "Dee", "answers": [2, 1, 2, 2, 3, 2]},]
cutoffs = [2, 2, 2, 3, 3, 3]

print(people[0]["answers"])
```

```python
ana = (people[0]["answers"])
```

```python
count = 0
ana = (people[0]["answers"])
for i in range(6):
    if ana[i] >= cutoffs[i]:
        count = count + 1
print(count)
if count >= 4:
    print("Positive screen")
else: 
    print("Negative screen")
```
